## 4. Functions and errors
### 4.1 A function that calculates the monthly cash transfer (KSh) from the number of children under five

```python
def cash_transfer(under5, base=2000, per_child=500, cap=3500):
    """Monthly cash transfer in KSh for a household."""
    return min(cap, base + per_child * under5)

print(cash_transfer(1, 2500, 400, 4000))              # 2500
print(cash_transfer(3))              # 3500, capped
print(cash_transfer(2, base=2500))   # 3500, a default value overridden
```

### 4.2 `print` shows a value, `return` gives it back

```python
def show_transfer(under5):
    print(2000 + 500 * under5)

x = show_transfer(1)    # shows 2500
print(x)                # None: nothing was given back
```

```python
def get_transfer(under5):
    return 2000 + 500 * under5

x = get_transfer(1)     # shows nothing
print(x + 500)          # 3000
```

### 4.4 `try` and `except`: don't crash on bad data

```python
def income_per_person(hh):
    """Income per person, or None if household size is 0."""
    try:
        return hh["income"] / hh["size"]
    except ZeroDivisionError:
        return None

print(income_per_person(households[0]))                      # 1925.0
print(income_per_person({"income": 3000, "size": 0}))        # None
```

### 5.1 One way to write it
A function that calculates a vulnerability score for each household using Tumaini's rule, then ranks households from the most to the least vulnerable.

```python
def vulnerability_score(hh):
    """Score a household's vulnerability using Tumaini Relief's rule."""
    score = 0
    if hh["income"] == 0:
        score += 2
    if hh["water_km"] > 5:
        score += 2
    score += hh["under5"]
    if hh["disability"] == "Yes":
        score += 2
    if hh["head_gender"] == "Female":
        score += 1
    return score

ranked = sorted(households, key=vulnerability_score, reverse=True)
for hh in ranked:
    print(hh["id"], vulnerability_score(hh))
```
