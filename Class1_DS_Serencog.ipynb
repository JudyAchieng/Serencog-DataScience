## 2. Python basics, using one household
We use **HH0002**: Turkana, female head aged 44, 6 people, 1 child under 5, a member living with a disability,
income KSh 2,200, 7.1 km to water.

### 2.1 Variables: labelled boxes that store a value
Use lowercase with underscores, no spaces, and don't start with a number.

```python
household_id = "HH0002"
county = "Turkana"
household_size = 6
children_under_5 = 1
monthly_income_ksh = 2200
water_distance_km = 7.1
disability_member = "Yes"
```

### 2.2 Data types
| Type | Meaning | Example |
|---|---|---|
| `int` | Whole numbers | `household_size = 6` |
| `float` | Decimals | `water_distance_km = 7.1` |
| `str` | Text, in quotes | `county = "Turkana"` |
| `bool` | `True` or `False` | `is_priority = True` |
| `None` | No value yet | `phone_number = None` |

```python
print(type(household_size))       # int
print(type(water_distance_km))    # float
print(type(county))               # str
print(type(household_id))         # str: we never do maths with an ID
```

### 2.3 Arithmetic: income per person for one household

```python
income_per_person = monthly_income_ksh / household_size
print(income_per_person)          # 366.666...
```

```python
# 7 people share 3 food baskets
print(7 // 3)    # 2 people per basket (whole division)
print(7 % 3)     # 1 person left over (remainder)
print(2 ** 3)    # 8 (power)
```

### 2.4 f-strings turn numbers into readable reports
`:,.2f` adds comma separators and 2 decimal places.

```python
print(f"{household_id} in {county}: {household_size} people, "
      f"KSh {income_per_person:,.2f} per person per month")
```

### 2.5 Comparisons and booleans: a targeting rule
Tumaini Relief marks a household as **priority** if it is more than 5 km from water, **or** has a member living with a disability.

Compare with `==  !=  >  <  >=  <=`. Combine with `and` (both needed), `or` (either is enough) and `not`.

```python
far_from_water = water_distance_km > 5             # True
has_disability = disability_member == "Yes"        # True
is_priority = far_from_water or has_disability     # True
print(is_priority)
```

### 2.6 Strings: field data is messy

```python
raw_county = "  turkana "
clean_county = raw_county.strip().title()    # strip removes the spaces, title fixes the capitals
print(clean_county)                          # Turkana
print(clean_county == "Turkana")             # True
print(len(household_id))                     # 6 characters
print(household_id[2:])                      # "0002": from position 2 to the end
```

### 2.7 Converting types: data from forms often arrives as text
The next cell gives an error on purpose. Read the last line: it tells you what went wrong.

```python
income_text = "2200"
income_text + 500        # TypeError: can't add text and a number
```

```python
int(income_text) + 500   # 2700
```

### Common mistakes
- `=` stores a value, `==` compares two values
- Text needs quotes, numbers don't: `6`, not `"6"`
- Case matters: `"turkana" == "Turkana"` is `False`
- Running cells out of order gives `NameError: name '...' is not defined`

### Exercise 4: The priority rule for HH0001
HH0001 is 4.3 km from water and has no member living with a disability. Is it a priority household?

```python
# Exercise 4: your code here
```

```python
# Solution
water_distance_km = 4.3
disability_member = "No"

far_from_water = water_distance_km > 5          # 4.3 > 5 → False
has_disability = disability_member == "Yes"     # "No" == "Yes" → False
is_priority = far_from_water or has_disability  # False or False → False
print(is_priority)                              # False
```
